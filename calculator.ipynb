**pyplot** module:-

The pyplot module in Matplotlib is a high-level interface that makes it easy to create plots and charts in Python, similar to MATLAB’s plotting style.

It is typically imported as:

***Python code:-***

**import matplotlib.pyplot as plt**

*Key Features:-*

Provides functions to create and customize plots.
Works like a state machine — commands affect the current figure and axes.

*Supports multiple plot types:-*

 line, bar, scatter, histogram, pie, etc.
Allows adding labels, titles, legends, grids, and annotations.

```python
import matplotlib.pyplot as plt

# using list of python
x=[2023,2024,2025,2026]   # list of x-coordinate
y=[15,25,30,20]                # list of y-coordinate

plt.plot(x,y)
plt.show()
```

```python
# use numpy array for faster access

# FIRST WAY:-
x=np.array([2023,2024,2025,2026])
y1=np.array([15,25,30,20])
y2=np.array([17,23,38,5])
plt.plot(x,y1, marker="*",markersize=15,markerfacecolor="#BF0FFF",markeredgecolor="#46C793", linestyle="dashdot",linewidth="3",color="#C43952")    # use color picker site for color selection
plt.plot(x,y2)   # we can use same customizing features for second line but instead we can simply create a dictionary of features and can be used for every required line.
plt.show()

```

```python
# SECOND WAY:-
# 10th marks representation of sibling:-
line_style=dict(marker="*",
                markersize=15,
                markerfacecolor="#BF0FFF",
                markeredgecolor="#46C793",
                linestyle="dashdot",
                linewidth="3",
                )
x=np.array(["Simran","Simpi","Sonu","Sristy"])
y1=np.array([90,80,89,98])   # marks
y2=np.array([40,49,48,43])   # age
y3=np.array([39,78,55,93])


plt.plot(x,y1,**line_style)   # using esterich to unpack the dictionary
plt.plot(x,y2,color="red",**line_style)
plt.plot(x,y3,color="cyan",**line_style)
```

Tick labels:-

Matplotlib has the ability to customize ticks and tick labels on axes, which enhances the readability and interpretability of graphs. This article will explore setting ticks and tick labels, providing a clear example to illustrate the core concepts.

# **What are Ticks?**

Ticks are markers on the axes of a plot that denote specific data points. By default, Matplotlib automatically generates these ticks based on the data being plotted. However, in many cases, users may want to customize these ticks to better reflect the data or improve clarity.

# **What are Tick Labels?**

Tick labels are the textual representations associated with each tick. They provide context for what each tick represents in terms of data values. Customizing tick labels can make a plot more informative and easier to understand.

```python

x = np.arange(0, 10, 1)
y = np.sin(x)

plt.plot(x, y)
plt.xticks([0, 2, 4, 6, 8], ['Zero', 'Two', 'Four', 'Six', 'Eight'])
plt.yticks([-1, 0, 1], ['Min', 'Zero', 'Max'])
plt.title('Sine Wave',fontweight="bold",fontsize=25,color="Red")
plt.show()
```


> Customizing Tick Properties with tick_params()

For further customization, you can use the tick_params() method. This function allows you to modify the appearance of the ticks, such as their size, color, direction, and visibility.

```python
x = [0, 1, 2, 3, 4]
y = [0, 1, 4, 9, 16]

plt.plot(x, y)
plt.tick_params(axis='x', labelsize=12, colors='red')

plt.show()
```

# Grid Lines⚓
a grid is a set of horizontal and/or vertical lines that make it easier to read and interpret data points on a plot.
You can add and customize grids using plt.grid().



```python
import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y, marker='o')

# Add grid lines
plt.grid()  # Default: both x and y axes

plt.title("Basic Grid Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
#Customizing the Grid

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y, marker='o')

# Custom grid
plt.grid(
    visible=True,       # Show grid
    which='both',       # 'major', 'minor', or 'both'
    axis='both',        # 'x', 'y', or 'both'
    color='gray',       # Grid line color
    linestyle='--',     # '--', '-.', ':', '-'
    linewidth=0.7,      # Line thickness
    alpha=0.7           # Transparency
)
plt.tick_params(axis='x', labelsize=12, colors='red')
plt.minorticks_on()  # Enable minor ticks for finer grid
plt.title("Customized Grid Example")
plt.show()
```

```python
x=[1,2,3,4,5]
y=[5,10,15,20,25]
plt.grid(axis="x",linewidth=3,color="gray",linestyle="dashed")
plt.grid(axis="y",linewidth=2)
plt.plot(x,y)
plt.show()
```

# Bar Charts♒

# Pie Chart⏰

```python
categories=np.array(["Freshman","Sophomores","Juniors","Seniors"])
values=np.array([300,250,275,225])
colors=np.array(["red","yellow","blue","green"])

plt.pie(values,labels=categories,
                autopct="%1.1f%%",  # % → start of a format specifier (like in C/Python string formatting)//  1.1f → floating-point number with 1 digit before the decimal and 1 digit after the decimal (e.g., 12.3)//  However, in a pie chart, autopct usually needs a percentage sign if you want to show percentages.
                colors=colors,
                explode=[0,0,0.1,0],    # In Matplotlib, the explode parameter in plt.pie() is used to offset one or more slices of a pie chart away from the center, making them stand out,   A list or tuple of floats, where each value represents the fraction of the radius to offset that slice
                shadow=True,
                startangle=180)


plt.title("")
plt.show()
```

# Scatter plots⚛
shows the relationship between two variables helps to identify a correlation(+,-,None) ex-study hours vs Test score.

```python
study_hours1=[0,1,1,2,3,4,5,6,7,7,8]
test_scores1=[55,60,65,62,68,70,75,78,82,85,87]

plt.scatter(study_hours1,test_scores1,color="red",
                                    alpha=0.4,   #transparency percentage
                                    s=200,       #size
                                    label="Class A")
study_hours2=[0,1,2,2,3,4,5,6,7,8,8]
test_scores2=[50,58,65,70,72,78,83,88,92,95,97]

plt.scatter(study_hours2,test_scores2,color="green",
                                    alpha=0.5,   #transparency percentage
                                    s=200,       #size
                                    label="Class B")

plt.title("Student score Report")
plt.xlabel("Hours Studied")
plt.ylabel("Test scores")

plt.legend()   # this will identify class A as red dot and class B as blue dot
plt.show()
```

# Histogram⛏
A Visuals representation of the distribution of quantitative data. They group values into bins(intervals) and counts how many fall in each group.

```python
scores=np.random.normal(loc=80,scale=10,size=100)  # normal distribution
# just to ensure that no scores go below zero or above 100, we can use the clip fxn
scores=np.clip(scores,0,100)  # reassign in the scores variable
plt.hist(scores,bins=10,  # for 10 bars
                color="lightgreen",
                edgecolor="black")

plt.title("Exam score")
plt.xlabel("SCore------->")
plt.ylabel("No. of Students------->")
plt.show()
```
